```python
import findspark
findspark.init()
from pyspark.sql import SparkSession

spark = SparkSession.builder.master("local[*]").getOrCreate()
sc = spark.sparkContext
```

```python
df=spark.read.csv('power.csv',inferSchema=True, header=True)
```

```python
##Выборки
```

```python
##Выбор определенных колонок
```

```python
df_lim=df.select('country','year','quantity')
```

```python
df_lim.take(5)
```

```python
df.take(5)
```

```python
##Переименовать колонки
```

```python
df_lim=df_lim.withColumnRenamed('year','y')
```

```python
df_lim.columns
```

```python
##Описание и статистика
```

```python
##Просмотреть содержание
```

```python
df_lim.show()
```

```python
##Посмотреть основную статистику
```

```python
df_lim.describe().show()
```

```python
##Сохранение
```

```python
df_lim.write.format('com.databricks.spark.csv').option('header','true').save('df_lim.csv')
```

```python
##Вычисления
```

```python
df=df_lim.withColumn('quantity x 2', df_lim['quantity']*2)
```

```python
df.take(5)
```

```python
df_lim.take(5)
```

```python
##Группировка
```

```python
df_grp=df.groupBy('y')
```

```python
df_sum=df_grp.sum('quantity')
```

```python
df_sum.show()
```

```python
##Сводные таблицы
```

```python
df.groupby('country').pivot('y').sum('quantity').show()
```
